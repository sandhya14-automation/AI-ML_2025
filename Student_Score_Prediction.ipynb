**Problem Statement:Given data: build a machine learning model that can predict Student Test Scores based on Hours Studied**

```python
#Step 1: Import Libraries
#munpy: For numerical computations.
#pandas: For handling datasets.
#sklearn.linear_model: For building the regression model.
#matplotlib.pyplot: For visualization.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#Step 2: Load and read the csv data file
```

```python
#Load and read the dataset
df = pd.read_csv('study_scores.csv')

#Display the dataset
print(df)
```

```python
#Step 3: Visualizating the Data
```

```python
#plt.scatter() is used to create a scatter plot to visualize how the scores varies with hours studied by the student.
#This helps to understand the relationship between hours studied by the student and scores achieved before using Linear Regression.
```

```python
plt.scatter(df['hours_studied'],df['score'],color='red',marker='+')
plt.xlabel('Hours studied')
plt.ylabel('Score')
plt.title('Score Prediction')
plt.show()
```

```python
#Step 4: Prepare Data for the Model
```

```python
#x is the independent variable(hours_studied).
#y is the dependent variable(score).
#We reshape x into a 2D DataFrame because Scikit-learn requires input features in this format.
```

```python
#Features (x) - Independent variable
x=df[['hours_studied']] #Feature should be in a 2D format

#Target Variable (y) - Dependent Variable
y=df['score']
```

```python
#Step 5: Create a model & Train that model
```

```python
#LinearRegression() creates a linear regression model.
#model.fit(x,y) trains the model using x(hours_studied) and y(score).
#The model learns the relationship between hours studied and score.
```

```python
#Create a Linear Regression model
model = LinearRegression()

#Train the model using the data
model.fit(x,y)
```

```python
#Step 6: Check Model Parameters
```

```python
#The model follows the equation y = b0 + b1 * x.
#intercept_ (b0) is the base score when hours studied = 0.
#coef_ (b1) is the score increase per hour studied.
```

```python
print('Intercept (b0):', model.intercept_)
print('Coefficient (b1):', model.coef_[0])
```

```python
#Step 7: Predict score for a student who studied 7.2 hours
```

```python
predict_score = model.predict([[7.2]])
print("Predicted Score for 7.2 hours:", predict_score[0])
```

```python
#Step 8: Plot Regression Line
```

```python
plt.scatter(x,y,color='red',marker='+',label='Actual Data')
plt.plot(x,model.predict(df[['hours_studied']]), color='blue',label='Regression Line')
plt.xlabel('Hours Studied')
plt.ylabel('Score')
plt.title('Linear Regression Score Prediction')
plt.legend()
plt.show()
```
